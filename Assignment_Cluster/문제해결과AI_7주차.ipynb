```python
def battle():
    Php = 80
    dmg = 30
    Php -= dmg
    print(f"전투중 HP: {Php}")

Php = 100
battle()
print(f"전투 밖 HP: {Php}")
```

```python
score = 0

def defeatMonster(monster):
    global score

    if monster == "슬라임":
        earnedScore = 100
    elif monster == "고블린":
        earnedScore = 200
    else:
        earnedScore = 500

    score += earnedScore
    print(f"{monster}처치! {earnedScore}점!")

defeatMonster("슬라임")
defeatMonster("고블린")
defeatMonster("드래곤")

print(f"최종 점수: {score}점")
```

```python
import random

hp = 100
monster_hp = 80
cnt = 3
def showStatus():
    print(f"플레이어 HP: {hp}")
    print(f"몬스터 HP: {monster_hp}")

def skillAttack():
    global monster_hp
    global cnt
    dmg = 30
    monster_hp -= dmg
    print(f"스킬공격! 데미지: {dmg}")
    cnt -= 1

def usePotion():
    global hp
    global cnt

    if cnt > 0:
        print(f"아직 사용할 수 없습니다! {cnt}턴 뒤에 사용하세요!")
    else:
        hp += 30
        print("포션 사용! HP30회복")
            

def monsterAtk():
    global hp
    dmg = random.randint(5,25)
    hp -= dmg
    print(f"몬스터 반격! 데미지: {dmg}")

print("몬스터가 나타났습니다!")

while hp > 0 and monster_hp > 0:
    print()
    showStatus()
    print("1. 스킬공격")
    print("2. 포션사용")
    action = input("행동 선택: ")

    if action == "1":
        skillAttack()
    elif action == "2":
        usePotion()
    else:
        print("없는 행동입니다. 다시 입력하세요!")
        continue

    if monster_hp > 0:
        monsterAtk()

    if hp > 0:
        print("몬스터 처치! 승리!")
    else:
        print("플레이어 패배!")
```
