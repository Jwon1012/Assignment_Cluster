```python
#실습I
import numpy as np

u = np.array([4,1,1])
v = np.array([2,1,0])

u1, u2, u3 = u
v1, v2, v3 = v

#행렬식 전개 부호 + - +
c1 = u2*v3 - u3*v2
c2 = -(u1*v3-u3*v1)
c3 = u1*v2 - u2*v1

result=np.array([c1,c2,c3])

print(f"i,j,k성분 = {c1},{c2},{c3}")
print(f"u x v = {result}")

ref = np.cross(u,v) #cf. 내적: dot
print(f"Np 결과: {ref}")
print(f"결과 확인: {np.allclose(result, ref)}")

#두 입력 벡터가 수직인지 확인
print(f"u . (u x v): {np.dot(u,result)}")
print(f"v . (u x v): {np.dot(v,result)}")
```

```python
#실습 II
import numpy as np

u = np.array([3,-1,2])
v = np.array([1,-2,4])

u1,u2,u3 = u
v1,v2,v3 = v

c1 = u2*v3 - u3*v2
c2 = -(u1*v3 - u3*v1)
c3 = u1*v2 - u2*v1

result = np.array([c1,c2,c3])

print(f"i, j, k 성분 = {c1},{c2},{c3}")
print(f"u x v = {result}")

#Np로 검산
print(f"Np 검산 결과: {np.cross(u,v)}")

#두 입력벡터가 수직인지 확인
print(f"u . (u x v) = {np.dot(u,result)}")
print(f"v . (u x v) = {np.dot(v,result)}")
```

```python
#실습 III
import numpy as np
u = np.array([3, 1, 0])
v = np.array([1, 3, 2])

#외적 계산
c_p = np.cross(u,v)

#넓이 = 외적 벡터의 길이
s_l = np.dot(c_p,c_p)
area = np.linalg.norm(c_p)

print(f"u x v = {c_p}")
print(f"넓이의 제곱 = {s_l}")
print(f"평행사변형의 넓이 = {area:.6f}")
```

```python
#실습 III
import numpy as np

#삼각형의 꼭짓점
A = np.array([-1, 2, 0])
B = np.array([2, 1, 3])
C = np.array([1, 1, -1])

#같은 점 A에서 출발하는 벡터
u = B - A
v = C - A

#외적 계산
c_p = np.cross(u,v)

#외적의 길이 = 평행사변형 길이
p_a = np.linalg.norm(c_p)

#삼각형의 넓이는 그거의 절반
triangle_area = p_a/2

print(f"u = {u}")
print(f"v = {v}")
print(f"u x v = {c_p}")
print(f"평행사변형의 넓이 = {(2*triangle_area):.6f}")
print(f"삼각형의 넓이 = {triangle_area:.6f}")
```

```python
#실습 IV
import numpy as np

u = np.array([4, 1, 1])
v = np.array([2, 1, 0])
w = np.array([0, 2, 3])

#외적 & 밑면의 넓이
c_p = np.cross(u, v)
S = np.linalg.norm(c_p)

#단위 법선벡터 & 수직 높이
unit_normal = c_p/S 
h = abs(np.dot(w,unit_normal))

#밑면 넓이 x 수직 높이
V1 = S * h

#스칼라 삼중곱의 절댓값
V2 = abs(np.dot(c_p,w))

#행렬식의 절댓값으로 계산
A = np.vstack([u,v,w])
V3 = abs(np.linalg.norm(A))

print(f"u x v = {c_p}")
print(f"밑면 넓이 S = {S:.6f}, 수직 높이 h = {h:.6f}")
print(f"체적 by S*h = {V1:.6f}")
print(f"체적 by 삼중곱 = {V2:.6f}")
print(f"체적 by Det = {V3:.6f}")
```

```python
#실습 V
import numpy as np

u = np.array([2, 5, -1])
v = np.array([3, 2, -3])

#같은 성분끼리 곱한걸 더하기
products = u * v
dot_manual = np.sum(products)

#Np함수로 내적 계산
dot_uv = np.dot(u,v)
dot_vu = np.dot(v,u)

#열벡터의 전치&행렬 곱으로 계산
u_col = u.reshape(3,1)
v_col = v.reshape(3,1)
dot_matrix=(u_col.T @ v_col).item()

print(f"성분별 곱 = {products}, 성분별 곱의 합 = {dot_manual}")
print(f"u.v = {dot_uv}, v.u = {dot_vu}")
print(f"uTv = {dot_matrix}, 교환법칙 확인 = {bool(dot_uv == dot_vu)}")

```

```python
#실습 VI
import numpy as np

u = np.array([2, -4])
v = np.array([4, 2])

#내적 계산
dot_uv = np.dot(u,v)
print(f"u.v = {dot_uv}")

#직교 여부 판단
print("두 벡터는 서로 직교합니다\n") if np.isclose(dot_uv,0) else print("두 벡터는 서로 직교하지 않습니다.\n")

#사이각 계산 # np.dot(u,v) = ||u||*||v||*cos(theta)
norm_product = np.linalg.norm(u) * np.linalg.norm(v)
theta = np.degrees(np.arccos(np.clip(dot_uv/norm_product, -1, 1))) if norm_product > 0 else print("0벡터가 포함되면 사이각은 정의되지 않습니다.")
print(f"두 벡터의 사이각 = {theta:.2f}도")
```
