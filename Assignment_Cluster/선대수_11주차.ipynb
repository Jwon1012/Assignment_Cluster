```python
#실습 7-1A 고유값과 고유벡터

import numpy as np
np.set_printoptions(precision = 4, suppress=True)

#행렬 A & 벡터 x
A = np.array([
    [1,2],
    [4,3]
],dtype = float)

x = np.array([1,2], dtype=float)

#전체벡터와 고유값 계산

Eval, Evtr = np.linalg.eig(A)
Eval = np.real_if_close(Eval)
Evtr = np.real_if_close(Evtr) #필요도 없는 허수부 제거
#고유값을 내림차순으로 정렬
order = np.argsort(Eval)[::-1]
print("고유값과 고유벡터")

for i in order:
    vector = Evtr[:, i]

    #첫번째 0이 아닌 성분을 1로
    first = np.flatnonzero(np.abs(vector) > 1e-12)[0]
    vector /= vector[first]

    print(f"L = {Eval[i]:g}, 고유벡터 = {vector}")

#검산식

Ax = A @ x

print("\nx =", x)
print("Ax =", Ax)
print("5x = ",5 * x)
print("Is Ax = 5x?", np.allclose(Ax, 5 * x))
```

```python
#실습 7-1B 고유값과 3차원 고유벡터 
import numpy as np
from fractions import Fraction
from math import lcm
np.set_printoptions(precision=4, suppress=True)

A = np.array([
    [3, -1, -1],
    [-12, 0, 5],
    [4, -2, -1]
])

#특성 다항식 계산
coefficients = (-1)**len(A)*np.poly(A)
print("특성 다항식 det(A-LI)")
print(np.poly1d(np.round(coefficients, 10), variable = "L"))

#고유값 & 고유벡터 계산
Eval, Evtr = np.linalg.eig(A)
Eval = np.real_if_close(Eval)
Evtr = np.real_if_close(Evtr) #필요도 없는 허수부 제거

for i in np.argsort(Eval):
    Ev = Eval[i]
    vector = Evtr[:, i]

    #이 문제의 고유벡터를 보기좋은 정수비로 전환
    vector /= vector[np.argmax(np.abs(vector))]
    fractions = [
        Fraction(float(t)).limit_denominator(1000)
        for t in vector
    ]
    common = lcm(*(f.denominator for f in fractions))
    x = np.array([int(f*common) for f in fractions])

    #첫번째 0이 아닌 성분을 양수로 통일

    if x[np.flatnonzero(x)[0]] < 0:
        x = -x
    
    print(f"\n고유값 L = {Ev:g}")
    print("고유벡터 x = ", x)
    print("Ax = ", A @ x)
    print("Ax = Lx 확인: ", np.allclose(A @ x, Eval * x))
```

```python
#실습 7-3 MATLAB -> Python

import numpy as np
A = np.array([
    [3, -1, -1],
    [-12, 0, 5],
    [4, -2, -1]
],dtype = float)

Evals, X = np.linalg.eig(A)
Evals = np.real_if_close(Evals)
X = np.real_if_close(X)

#고유값을 2, 1, -1순으로 정렬
order = np.argsort(Evals)[::-1]
Evals = Evals[order]
X = X[:, order]

#고유값 -> 대각행렬
lamda = np.diag(Evals)
np.set_printoptions(precision=4, suppress=True, floatmode="fixed")

print("X =")
print(X)
print("\nlamda = ")
print(lamda)
```
