```python
#실습I

import numpy as np

A = np.array([
    [1, 2, 3],
    [2, 5, 3],
    [1, 0, 8]
], dtype = float)

I = np.eye(3)

AI = np.hstack((A, I))

def show_matrix(operation):
  """행 연산 과정 출력"""
  print(f"\n{operation}")
  print(AI)

print(f"초기 첨가행렬: [A | I]")
print(AI)

#1단계: 첫번째 열에서 1행 아래의 원소를 0으로
AI[1] = -2 * AI[0] + AI[1]
show_matrix("R2 <- (-2)R1 + R2")

AI[2] = (-1) * AI[0] + AI[2]
show_matrix("R3 <- (-1)R1 + R3")

#2단계: 두번째 열에서 피벗 아래의 원소를 0으로
AI[2] = 2 * AI[1] + AI[2]
show_matrix("R3 <- (2)R2 + R3")

#3단계 : 세번째 피벗을 1로
AI[2] = (-1)*AI[2]
show_matrix("R3 <- (-1)R3")

#4단계: 세번째 열에서 피벗 위의 원소를 0으로
AI[1] = AI[1] + 3*AI[2]
show_matrix("R2 <- R2 + (3)R3")

AI[0] = AI[0] - 3*AI[2]
show_matrix("R1 <- R1 - (3)R3")

#5단계: 두번째 열에서 피벗 위의 원소를 0으로
AI[0] = AI[0] - 2*AI[1]
show_matrix("R1 <- R1 - (2)R2")

#오른쪽 부분이 A의 역행렬
A_inv = AI[:, 3:]
print("\n최종 첨가행렬 [I | A^(-1)]")
print(AI)

print("\nA의 역행렬")
print(A_inv)

print("\n검산: A * A^(-1)")
print(A @ A_inv)
```

```python
#실습II

import numpy as np
np.set_printoptions(suppress=True, precision=2)

A = np.array([
    [1, -3, 4],
    [2, -5, 7],
    [0, -1, 1]
],dtype=float)

I = np.eye(3)

AI=np.hstack((A, I))

def show_matrix(title):
  """첨가행렬을 [A | I]형태로 출력"""
  print(f"\n{title}")

  for row in AI:
    left = " ".join(f"{value:6.1f}" for value in row[:3])
    right = " ".join(f"{value:6.1f}" for value in row[3:])
    print(f"[{left} | {right}]")

show_matrix("초기 첨가행렬 [A | I]")

#1단계: R2 <- (-2)R1 + R2
AI[1] = (-2)*AI[0] + AI[1]
show_matrix("1단계: R2 <- (-2)R1 + R2")

#2단계: R3 <- R2 + R3
AI[2] = AI[1] + AI[2]
show_matrix("2단계: R3 <- R2 + R3")

#3단계: R1 <- 3R2 + R1
AI[0] = 3 * AI[1] + AI[0]
show_matrix("3단계: R1 <- 3R2 + R1")

#왼쪽 A 부분과 오른쪽 부분 분리
left_mat = AI[:, :3]
right_mat = AI[:, 3:]

print("\n행 연산 후의 왼쪽 행렬")
print(left_mat)
print("\n행 연산 후의 오른쪽 행렬")
print(right_mat)

#역행렬 존재 여부 확인

if np.linalg.matrix_rank(A) < A.shape[0]:
  print("\n판정 결과, A의 행들이 선형종속입니다.")
  print("\n왼쪽 행렬에 모든 원소가 0인 행이 발생했습니다.")
  print("\n따라서 A를 단위행렬로 만들 수 없고 A의 역행렬은 존재하지 않습니다.")
else:
  A_inv = right_mat
  print("\nA의 역행렬")
  print(A_inv)

det_A = np.linalg.det(A)
print("|A| =", round(det_A))
```

```python
#실습III

import numpy as np
A = np.array([
    [1,2,3],
    [2,5,3],
    [1,0,8]
], dtype = float)

#소행렬
def get_minor(matrix, row, col):
  return np.delete(
      np.delete(matrix, row, axis=0),
  col,
  axis = 1
  )

#여인수행렬 계산
def cofactor_matrix(matrix):
  n = matrix.shape[0]
  C = np.zeros((n,n))
  for i in range(n):
    for j in range(n):
      minor = get_minor(matrix, i, j)
      C[i, j] = ((-1) ** (i + j)) * np.linalg.det(minor)
  return C

#행렬식 계산
det_A = round(np.linalg.det(A))

if det_A == 0:
  print("행렬식이 0이므로 역행렬이 존재하지 않습니다.")
else:
  #1. 여인수 행렬
  C = cofactor_matrix(A)
  #2. 수반 = 여인수 전치
  adj_A = C.T
  #3. 역행렬
  inv_A = (1 / det_A) * (adj_A)
  #부동소수점 오차 정리
  C = np.round(C, 10)
  adj_A = np.round(adj_A, 10)
  inv_A = np.round(inv_A, 10)

print("행렬 A")
print(A)
print("\n행렬식 det(A)")
print(det_A)
print("\n여인수행렬 C")
print(C)
print("\n수반행렬 adj(A)")
print(adj_A)
print("\n역행렬 A^(-1)")
print(inv_A)
print("\n검산 A * A^-1")
print(np.round(A @ inv_A, 10))
```

```python
#실습 IV

import numpy as np
np.set_printoptions(suppress=True, precision=2)

#계수행렬
A = np.array([
    [1,2,3],
    [1,3,6],
    [2,6,13]
], dtype = float)

#상수벡터 b
b = np.array([      # x + 2y + 3z = 1
    [1],            # x + 3y + 6z = 3
    [3],            # 2x + 6y + 13z = 5
    [5]             # 위의 선형방정식의 해를 구하는 프로그램
], dtype = float)

print("계수행렬 A")
print(A)
print("\n상수벡터 b")
print(b)

#행렬식 계산
det_A = np.linalg.det(A)
print("\n행렬식 det(A)")
print(round(det_A, 10))


if np.isclose(det_A, 0):
  print("행렬식이 0이므로 역행렬이 존재하지 않습니다.")
else:
  #A의 역행렬
  A_inv = np.linalg.inv(A)

  print("\nA의 역행렬")
  print(A_inv)
  X = A_inv @ b
  print("\n선형시스템의 해")
  print(X)

  x, y, z = X.flatten()
  print(f"\nx = {x:.0f}")
  print(f"\ny = {y:.0f}")
  print(f"\nz = {z:.0f}")

  print("\n검산 A * X")
  print(A @ X)
```

```python
#실습 V
import numpy as np


#2x2 행렬의 행렬식 계산 함수
def determinant_2x2(matrix):
  return (
      matrix[0, 0] * matrix[1, 1] -
      matrix[0, 1] * matrix[1, 0]
  )

#계수행렬 (넘파이 배열로 생성)
A = np.array([
    [3, -2],
    [-5, 4]
])

#상수벡터 (넘파이 배열로 생성)
b = np.array([6, 8])

#A1: A의 첫째 열 -> b
A1 = np.array([
    [b[0], A[0, 1]],
    [b[1], A[1, 1]]
])

#A2: A의 둘째 열 -> b
A2 = np.array([
    [A[0, 0], b[0]],
    [A[1, 0], b[1]]
])

#각각의 D계산
D = determinant_2x2(A)
D1 = determinant_2x2(A1)
D2 = determinant_2x2(A2)

print("계수행렬 A=\n", A)
print("\nA1 = \n", A1)
print("\nA2 = \n", A2)
print("\ndet(A) = ", D)
print("det(A1) = ", D1)
print("det(A2) = ", D2)

#크레머의 규칙 적용
if D == 0:
  print("\ndet(A)가 0이므로 못써요.")
else:
  x1 = D1 / D
  x2 = D2 / D
  print("\n크레머의 규칙 계산")
  print(f"x1 = det(A1) / det(A) = {D1} / {D} = {x1}")
  print(f"x2 = det(A2) / det(A) = {D2} / {D} = {x2}")
  print("\n선형시스템의 해")
  print(f"x1 = {x1}")
  print(f"x2 = {x2}")

  #검산
  result1 = 3 * x1 - 2 * x2
  result2 = -5 * x1 + 4 * x2
  print("\n검산")
  print(f"3x1 - 2x2 = {result1}")
  print(f"-5x1 + 4x2 = {result2}")
```

```python
#실습 VI

import numpy as np
np.set_printoptions(suppress=True, precision=2)

#계수행렬 A
A = np.array([
    [1, -1, 1],
    [1, 1, -2],
    [1, 2, 1]
], dtype=float)

#상수벡터 B
b = np.array([0, 1, 6], dtype=float)
#기준행렬식 D = det(A)
D = round(np.linalg.det(A))

print(f"계수행렬 A\n{A}\n상수벡터 b\n{b}")
print(f"\nD = det(A) = {D}")

if D == 0:
  print("행렬식이 0이므로 크래머의 규칙을 적용할 수 없습니다.")
else:
  determinants=[]
  solution=[]

  #A의 각 열을 b로
  for i in range(A.shape[1]):
    Ai = A.copy()
    Ai[:, i] = b

    Di = round(np.linalg.det(Ai))
    xi = Di / D
    determinants.append(Di)
    solution.append(xi)

    print(f"\nA{i + 1}: {i + 1}번째 열을 b로 교체")
    print(Ai)
    print(f"D{i + 1} = det(A{i + 1}) = {Di}")
    print(
        f"x{i + 1} = D{i + 1}/D"
        f"= {Di}/{D} = {xi}"
    )
solution = np.array(solution)

print("\n최종 해")

for i,value in enumerate(solution, start=1):
  print(f"x{i} = {value}")

#검산
print(f"\nA의 검산결과 A @ x {A @ solution}")
print(f"\n원래 상수벡터 b {b}")
```

```python
#실습 VII

import math

def determinant_2x2(matrix):
  return (
      matrix[0][0] * matrix[1][1]
      - matrix[0][1] * matrix[1][0]
  )

#각도를 라디안으로 변환
angle_25 = math.radians(25)
angle_15 = math.radians(15)

#계수행렬 A
A = [
    [math.cos(angle_25), -math.cos(angle_15)],
    [math.sin(angle_25), math.sin(angle_15)]
]

#상수벡터 b
b = [0, 30]

#A1: 첫번째 열을 b로
A1 = [
    [b[0], A[0][1]],
    [b[1], A[1][1]]
]

#A2: 두번째 열을 b로

A2 = [
    [A[0][0], b[0]],
    [A[1][0], b[1]]
]

#행렬식 계산
D = determinant_2x2(A)
D1 = determinant_2x2(A1)
D2 = determinant_2x2(A2)

print("계수행렬 A")
for row in A:
  print(row)

print("\nA1: 첫번째 열을 b로")
for row in A1:
  print(row)

print("\nA2: 두번째 열을 b로")
for row in A2:
  print(row)
print(f"\nD = det(A) = {D: .6f}")
print(f"\nD1 = det(A1) = {D1: .6f}")
print(f"\nD2 = det(A2) = {D2: .6f}")

#크래머의 규칙

if math.isclose(D, 0.0, abs_tol=1e-12):
  print("\ndet(A)가 0이므로 유일한 해가 없습니다.")
else:
  T1 = D1/D
  T2 = D2/D
  print("\n크래머의 규칙 계산\n")
  print(f"T1 = D1/D = {D1: .6f}/{D: .6f}")
  print(f"T1 = {T1: .4f}")
  print(f"\nT2 = D2/D = {D2: .6f}/{D: .6f}")
  print(f"T2 = {T2: .4f}")

  #검산
  equation1 = (
      math.cos(angle_25) * T1
      - math.cos(angle_15) * T2
  )

  equation2 = (
      math.sin(angle_25) * T1
      + math.sin(angle_15) * T2
  )

  print("\n검산")
  print(f"수평 방향 합 = {equation1:.6f}")
  print(f"수직 방향 합 = {equation2:.6f}")
```
