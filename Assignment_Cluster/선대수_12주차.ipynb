```python
#실습 I 두 벡터의 내적

import numpy as np
u= np.array([[2],[-5],[-1]])
v = np.array([[3], [2], [-3]])

# .T: 전치, @: 행렬곱, .item(): 숫자 하나 추출
u_dot_v = (u.T @ v).item()
v_dot_u = (v.T @ u).item()

print("u . v", u_dot_v) #-1
print("v . u", v_dot_u) #-1
print("교환법칙 확인:", u_dot_v == v_dot_u) #True
```

```python
#실습 II 세 벡터의 내적
import numpy as np

u = np.array([1,2,4])
v = np.array([2, -3, 5])
w = np.array([4, 2, -3])

print("u . v = ", np.dot(u, v))
print("u. w = ", np.dot(u, w))
print("v. w = ", np.dot(v, w))
```

```python
#실습 III 노름, 벡터의 거리
import numpy as np
u = np.array([1,2,3])
v = np.array([-4,3,5])

#각 벡터의 길이
N_u = np.linalg.norm(u)
N_v = np.linalg.norm(v)

#두 벡터 사이의 길이
difference = u-v
distance = np.linalg.norm(u-v)

print(f"u의 길이: {N_u:.6f}")
print(f"v의 길이: {N_v:.6f}")
print("v - u = ", difference)
print("각 성분의 제곱: ", difference ** 2)
print(f"두 벡터의 길이: {distance:.6f}")
```

```python
#실습 IV 노름 & 내적
u = np.array([2,-5,-1])
v = np.array([-7,-4,6])

#1. 내적
d_p = np.dot(u, v) #내적 0 -> 벡터가 수직 ∵각도의 코사인값 = 0

#2. 길이의제곱
norm_u_squared = np.dot(u, u)

#3. 거리의 제곱
difference = u - v
distance_squared = np.dot(difference, difference)

print("u . v = ", d_p)
print("||u||^2 = ", norm_u_squared)
print("||u-v||^2 = ", distance_squared)
```

```python
# 실습 V 벡터가 이루는 각 구하기
import numpy as np

u = np.array([1,1,0], dtype = float)
v = np.array([0,1,1], dtype = float)

d_p = np.dot(u,v)
n_u = np.linalg.norm(u)
n_v = np.linalg.norm(v)

d = np.linalg.norm(u - v)

print(f"내적: {d_p:.6f}")
print(f"거리: {d:.6f}")

#두 벡터 사이의 각도

if n_u == 0 or n_v == 0:
    print("영벡터가 있어서 각도를 계산할 수 없습니다.")
else:
    cos_theta = d_p/(n_u*n_v)
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta_rad = np.arccos(cos_theta)
    theta_deg = np.degrees(theta_rad)

    print(f"cos(theta) = {cos_theta:.1f}")
    print(f"각도 = {theta_deg:.0f}도 ")
```

```python
#실습 VI 벡터가 이루는 각 구하기 II
import numpy as np

u = np.array([1,-1,0], dtype = float)
v = np.array([2, -1, 2], dtype = float)

#내적과 벡터의 길이
d_p = np.dot(u,v)
n_u = np.linalg.norm(u)
n_v = np.linalg.norm(v)

if n_u == 0 or n_v == 0:
    raise ValueError ("영벡터가 포함되면 각도를 정의할 수 없어요")

#cos(theta) = 내적 / 두 벡터의 길이의 곱
cos_theta = d_p/(n_u*n_v)
cos_theta = np.clip(cos_theta, -1.0, 1.0)

#라디안으로 각도를 구한 뒤 도 단위로 변환
theta_rad = np.arccos(cos_theta)
theta_deg = np.degrees(theta_rad)

print(f"내적:            = {d_p:.6f}")
print(f"u의 길이:        = {n_u:.6f} ")
print(f"v의 길이:        = {n_v:.6f}")
print(f"cos(theta)      = {cos_theta:.6f}")
print(f"각도(rad)        = {theta_rad:.6f}")
print(f"각도(도)         = {theta_deg:.0f}")
```

```python
#실습 VII 벡터의 직교

import numpy as np
u = np.array([1, 2], dtype = float)
v = np.array([6, -3], dtype = float)

d_p = np.dot(u, v)
n_u = np.linalg.norm(u)
n_v = np.linalg.norm(v)

print(f"내적 = {d_p:.6f}")

if n_u == 0 or n_v == 0: # u,v가 0벡터가 아닌이상 벡터가 이루는 각이 수직이 되면 코사인 값이 0이 되어서 0이 나옴.
    print("직 to the 교.")
    print("but 영벡터가 있는 각도는 정의 X")
else:
    cos_theta = d_p/(n_u * n_v)
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta_deg = np.degrees(np.arccos(cos_theta))
    is_90_degrees = np.isclose(
        cos_theta, 0.0, atol=1e-10, rtol=0.0
    )

    print(f"cos(theta) = {cos_theta:.6f}")
    print(f"각도 = {theta_deg:.0f}도")
    print(f"직교합니다. " if is_90_degrees else "직교하지 않습니다.")
```
