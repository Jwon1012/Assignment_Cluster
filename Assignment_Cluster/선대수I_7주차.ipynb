```python
#실습I (가우스 소거법을 이용한 선형방정식 해법), 디버깅으로 재확인 예정
from fractions import Fraction

def print_matrix(M): ##코드의 핵심로직 
    '''첨가행렬 출력'''
    for row in M: ##?
        coefficients = " ".join(f"{str(x):>6}"for x in row[:-1]) 
        print(f"[{coefficients} | {str(row[-1]):>6}")
    print()

#문제의 첨가행렬: 마지막 열은 우변 상수

M = [
    [0, -1, -1, 1, 0],
    [1, 1, 1, 1, 6],
    [2, 4, 1, -2, -1],
    [3, 1, -2, 2, 3]
]

#정확한 분수 계산을 위한 자료형 변환 
M = [[Fraction(X) for X in row] for row in M] ##코드의 핵심로직 II
n = len(M)

print("초기 첨가행렬")
print_matrix(M)

for col in range(n):
    #I. 현재 열에서 0이 아닌 피벗 탐색
    pivot_row = next(
        (r for r in range(col,n) if M[r][col] != 0), None ##한줄 for문 & if문 짬뽕
    )

    if pivot_row is None:
        raise ValueError("유일한 해가 없습니다.")

    #II. 피벗이 0이면 아래 행과 교환
    if pivot_row != col:
        M[col], M[pivot_row] = M[pivot_row], M[col]
        print(f"R{col + 1} <-> R{pivot_row + 1}")
        print_matrix(M)

    #III. 피벗 행을 피벗 값으로 나누어 피벗을 1로
    pivot = M[col][col]
    M[col] = [x / pivot for x in M[col]]

    #IV. 피벗 위와 아래의 원소를 모두 0으로
    for r in range(n):
        if r == col:
            continue

        factor = M[r][col]

        if factor != 0:
            M[r] = [ ##가우스-조던 방법의 원리 파악
                M[r][c] - factor * M[col][c]
                for c in range(n + 1)
            ]

            print(
                f"R{r + 1} <- R{r + 1}"
                f"- ({factor}) x R{col + 1}"
            )

            print_matrix(M)

print("최종 해")
for i in range(n):
    print(f"x{i + 1} = {M[i][-1]}")
```

```python
#실습II (가우스 소거법을 이용한 선형방정식 해법 II)

from fractions import Fraction

def print_matrix(M):
    '''첨가행렬 출력'''
    for row in M:
        left = " ".join(f"{str(x):>5}"for x in row[:-1])
        print(f"[{left} | {str(row[-1]):>5}]")
    print()

#계수행렬 (마지막 열은 우변의 상수벡터)
M = [
    [1, -2, 1, 0],
    [0, 2, -8, 8],
    [-4, 5, 9, -9]
]

#분수를 정확히 계산하기 위한 변환
M = [[Fraction(x) for x in row]for row in M]

print("1. 초기 첨가행렬")
print_matrix(M)

#세번째 행의 x1계수를 0으로
M[2] = [M[2][j] + 4 * M[0][j] for j in range(4)]

print("2. R3 <- R3 + 4R1")
print_matrix(M)
#두번째 행의 피벗을 1로

M[1] = [(x/2) for x in M[1]]

print("3. R2 <- R2 / 2")
print_matrix(M)
#세 번째 행의 x2계수를 0으로
M[2] = [M[2][j] + 3 * M[1][j] for j in range(4)]

print("4. R3 <- R3 +3R2")
print_matrix(M)

#후진 대입:마지막 방정식부터 역순으로
n = 3
x = [Fraction(0) for _ in range(n)]

print("5. 후진 대입")

for i in range(n-1, -1, -1):
    known = sum(M[i][j] * x[j] for j in range(i+1, n))
    x[i] = (M[i][-1] - known) / M[i][i]
    print(f"x{i + 1} = {x[i]}")

print("\n최종 결과")
for i,value in enumerate(x, start=1): ##저놈은 또 뭐여
    print(f"x{i} = {value}")
```

```python
#실습III (같은 좌변, 다른 우변)

def print_matrix(M):
    '''계수 2열과 각 시스템의 상수항을 구분해 출력'''
    for row in M:
        print(
            f"[{row[0]:>3} {row[1]:>3}"
            f" | {row[2]:>3} | {row[3]:>3}]"
        )
    print()

#[x의 계수, y의계수, 첫번째 우변, 두번째 우변]

M = [
    [1, 2, 2, 1],
    [3, 7, 8, 7],
]

print("1. 초기 첨가행렬")
print_matrix(M)

#R2 <- R2 - 3R1

M[1] = [M[1][j] - 3 * M[0][j] for j in range(4)]
print("2. R2 <= R2 - 3R1")
print_matrix(M)

#R1 <- R1 - 2R2

M[0] = [M[0][j] - 2 * M[1][j] for j in range(4)]

print("3. R1 <- R1 - 2R2")
print_matrix(M)

#왼쪽의 단위행렬이 되었으니까 오른쪽 각 열이 해
for system in range(2):
    column = 2 + system
    x = M[0][column]
    y = M[1][column]

    print(f"시스템 {system + 1}의 해: x = {x}, y = {y}")
```

```python
#실습IV (Ax = b -> Ly = b & Ux = y), 디버깅으로 확인 예정
from fractions import Fraction

def print_matrix(name, M):
    '''행렬을 보기 좋게 출력'''
    print(name)
    for row in M:
        print("[" + " ".join(f"{str(x):>5}" for x in row) + "]")
    print()

A = [
    [1, 2, 1],
    [2, 3, 3],
    [-3, -10, 2],
]

n = len(A)

#U(상삼각행렬)는 A의 복사본에서 시작
#Fraction을 써서 분수를 정확히 계산

U = [[Fraction(x) for x in row] for row in A] 

#L(하삼각행렬)은 단위행렬에서 시작
L = [
    [Fraction(1 if i == j else 0) for j in range(n)] #3x3 단위행렬 생성 코드
    for i in range(n)
]

print_matrix("초기행렬 A", A)

#가우스 소거법: 피벗 아래의 원소를 0으로.

for k in range(n - 1):
    if U[k][k] == 0:
        raise ValueError("피벗이 0이므로 행교환이 필요합니다.")

    for i in range(k + 1, n):
        #소거 배수 = 없앨 원소 / 피벗
        multiplier = U[i][k] / U[k][k]

        #소거 배수를 L에 저장
        L[i][k] = multiplier

        #R1 <- Ri - multiplier x Rk
        U[i] = [
            U[i][j] - multiplier * U[k][j]
            for j in range(n)
        ]

        print(f"R{i + 1} <- R{i + 1} - ({multiplier}) x R{k + 1}")
        print_matrix("현재 U", U)

    print_matrix("최종 L", L)
    print_matrix("최종 U", U)

    #LU 계산
    LU = [
        [
            sum(L[i][k] * U[k][j] for k in range(n)) ##행렬곱(검산식) 계산 로직
            for j in range(n)
        ]
        for i in range(n)
    ]

    print_matrix("검산: LU",LU)
    print("A = LU확인", LU == A)

```

```python
#실습V (시험에는 X, 대신 LU로 쪼개는건 필요)

from fractions import Fraction

def print_matrix(name, M):
    '''행렬을 보기 좋게 출력'''
    print(name)
    for row in M:
        print("[" + " ".join(f"{str(x):>5}" for x in row) + "]")
    print()

#주어진 행렬
A = [
    [1, 2, 1],
    [2, 3, 3],
    [-3, -10, -2],
]

n = len(A)

#U는 A의 복사본에서, L은 단위행렬 I에서 시작.

U = [[Fraction(x) for x in row] for row in A]
L = [[Fraction(1 if i == j else 0) for j in range(n)] for i in range(n)]

print_matrix("초기행렬", A)

#가우스 소거법: 피벗(행렬의 중심축, 단위행렬의 1에 해당하는 부분) 아래의 원소를 0으로
for k in range(n-1):
    if U[k][k] == 0:
        raise ValueError("피벗이 0이므로 행교환이 필요합니다.")

    for i in range(k+1, n):
        #소거배수 = 없앨 원소/피벗
        multiplier = U[i][k]/U[k][k]
        #소거변수를 L에 저장
        L[i][k] = multiplier

        #Ri = Ri - multiplier x Rk
        U[i] = [U[i][j] - multiplier * U[k][j] for j in range(n)]

        print(f"R{i + 1} <- R{i + 1} - {multiplier} * R{k + 1}")
        print_matrix("현재 U", U)

    print_matrix("최종 상삼각행렬 U", U)
    print_matrix("최종 하삼각행렬 L", L)

    #행렬곱 LU

    LU = [
        [sum(L[i][m] * U[m][j] for m in range(n))
         for j in range(n)]
        for i in range(n)
    ]
    print_matrix("LU검산", LU)
    print("LU == A 확인 ", LU == A)

```

```python
#실습VI(시험에는 X, 대신 deg2rad 부분은 중요)
import numpy as np
import matplotlib.pyplot as plt

theta_deg = 60
theta = np.deg2rad(theta_deg) # degree를 radian 값으로 바꿔주는 코드

# 시계 방향 회전행렬
A = np.array([
    [np.cos(theta), np.sin(theta)],
    [-np.sin(theta), np.cos(theta)]
])

# 각 열이 하나의 꼭짓점인 2x4 행렬
# P1(1,1), P2(1,2), P3(2,2), P4(2,1)
P = np.array([
    [1, 1, 2, 2],
    [1, 2, 2, 1]
])

# 네 꼭짓점을 한방에 변환
Q = A @ P
np.set_printoptions(precision=4, suppress=True)

print("회전행렬 A:")
print(A)

print("\n각 꼭짓점의 변환 결과: ")
for i in range(P.shape[1]):
    print(
        f"P{i + 1}: ({P[0, i]:.0f}, {P[1, i]:.0f})"
        f" -> ({Q[0, i]:.4f}, {Q[1, i]:.4f})"
    )

# 원점에서 각 꼭짓점까지의 거리 보존 확인
b4 = np.linalg.norm(P, axis=0)
after = np.linalg.norm(Q, axis=0)

print("\n원점까지의 거리 보존:", np.allclose(b4, after))

# 첫 꼭짓점을 끝에 추가하여 닫힌 정사각형으로 그리기
P_closed = np.column_stack((P, P[:, 0]))
Q_closed = np.column_stack((Q, Q[:, 0]))

fig, ax = plt.subplots(figsize=(7, 7))

ax.plot(
    P_closed[0], P_closed[1],
    "o-", color="royalblue", label="Original"
)

ax.plot(
    Q_closed[0], Q_closed[1],
    "o--", color="darkorange",
    label=f"Rotated {theta_deg} degrees clockwise"
)

# 꼭짓점 이름 표시
for i in range(P.shape[1]):
    ax.annotate(
        f"P{i + 1}", P[:, i],
        xytext=(6, 8), textcoords="offset points"
    )
    ax.annotate(
        f"P{i + 1}'", Q[:, i],
        xytext=(6, -16), textcoords="offset points"
    )

# 원점에서 첫 꼭짓점으로 향하는 벡터 표시
for point, color in [(P[:, 0], "royalblue"), (Q[:, 0], "darkorange")]:
    ax.annotate(
        "", xy=point, xytext=(0, 0),
        arrowprops=dict(arrowstyle="->", color=color)
    )

ax.axhline(0, color="gray", linewidth=1)
ax.axvline(0, color="gray", linewidth=1) # 세로축 그리기로 수정
ax.scatter(0, 0, color="black", zorder=5, label="Origin")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Rotation about the origin")
ax.set_aspect("equal", adjustable="box")
ax.margins(0.2)
ax.grid(True, alpha=0.3) # grind -> grid 오타 수정
ax.legend()
plt.tight_layout()
plt.show() # () 추가
```
