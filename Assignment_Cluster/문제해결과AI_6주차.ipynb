```python
import random


number = random.sample(range(10),3)

while True:
    guess=input("추측한 세자리의 숫자를 입력하세요.")


    strike, ball, out = 0, 0, 0
    secret = ""
    for i in range(3):
        secret += str(number[i])

    for i in range(3):
        if secret[i] == guess[i]:
            strike += 1
        elif secret[i] in guess:
            ball += 1
        
    if strike==3:
        print("성공!!")
        break

    print(f"strike = {strike}, ball = {ball}, out = {out}")
```
